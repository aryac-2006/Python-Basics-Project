```python
import random
import math
# Create an empty set to store unique numbers
numbers = set()

try:
    # Take 10 numbers as input
    for i in range(10):
        num = int(input(f"Enter number {i+1}: "))
        numbers.add(num)

    # convert set into tuple
    numbers_tuple = tuple(numbers)
    print("\nUnique numbers: ", numbers_tuple)

    # Pick random numbers from tuple
    if len(numbers_tuple) >= 3:
        random_num = random.sample(numbers_tuple, 3)
        print("3 random numbers: ", random_num)
    else:
        print("Not enough to pick 3 random numbers.")

    # Find the sum of tuple elements
    total = sum(numbers_tuple)

    # print square root of sum
    if total >= 0:
        print("Square root of sum: ",math.sqrt(total))
    else:
        print("Square root cannot be calculated.")

except ValueError:
    print("Invalid input! please enter only numbers")
```
