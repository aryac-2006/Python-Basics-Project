```python
# Lambda function to calculate the square of a number
square = lambda x: x * x

try:
    # Generate numbers from 1 to 20
    numbers = range(1, 21)
    # Store square in a list
    square = [square(num) for num in numbers]

    # Print all squares
    print("Square: ")
    print(square)
    # Print only even numbers
    print("\nEven square: ")
    for num in square:
        if num % 2 == 0:
            print(num, end=" ")

except Exception as e:
    print("Error occured: ", e)
```
