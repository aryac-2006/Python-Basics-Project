```python
# Function to manage student records
def student_database():
    student = {}
    # Menu runs until user chooses exit
    while True:
        print("\n===Student database===")
        print("1. Add Students")
        print("2. Search student")
        print("3. Display All Students")
        print("4. Exit")

        try:
            # Take user's choice
            choice = int(input("Enter your choice: "))
            # Add a new student
            if choice == 1:
                roll = int(input("Enter roll: "))
                name = input("Enter name: ")
                age = int(input("Enter age: "))
                city = input("Enter city: ")

                student.update({roll: {"name": name,
                                       "roll": roll,
                                       "age": age,
                                       "city": city}
                                })
                print("Student added successfully.")
            # Add student details using update()
            elif choice == 2:
                roll = int(input("Enter roll: "))
                # Search student details using get()
                stu = student.get(roll)
                if stu:
                    print("\nStudent Found")
                    print("Name:", stu["name"])
                    print("Age:", stu["age"])
                    print("City:", stu["city"])
                else:
                    print("\nStudent not found")

           # Display all student records
            elif choice == 3:
                if len(student) == 0:
                    print("\nNo Student Record Found")
                else:
                    print("\nAll students:")
                    for roll, details in student.items():
                        print("-----------------------")
                        print("Roll:", roll)
                        print("Name:", details["name"])
                        print("Age:", details["age"])
                        print("City:", details["city"])
            # Exit the program
            elif choice == 4:
                print("Exiting program...")
                break
            else:
                print("\nInvalid Choice")

        except ValueError:
            print("\nPlease enter numbers where required")

student_database()
```
