```python
# Create analyze_string function
def analyze_string(s):
    # Length and reverse of string
    print("\nLength of string:", len(s))
    print("Reverse of string:", s[::-1])

    # Count and print how many vowels are in the string
    count = 0
    vowels = "aeiou"
    for ch in s.lower():
        if ch in vowels:
            count += 1
    print("Number of vowels:", count)

    # print each character with its positive and negative index.
    print("\nCharacter with Positive and Negative Index:")
    for i in range(len(s)):
       print(f"{i+1}. Character |  Positive index | Negative index")
       print("\t",s[i] ,"\t\t \t ", i  ," \t\t \t  ",  i - len(s) )

s = input("Enter a string: ")
# Check if the string is empty
if s == "":
    print("Empty string")
else:
    analyze_string(s)
```
