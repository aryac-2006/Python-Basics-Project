```python
def manage_marks():
    # Create an empty list to store marks
    marks = []
    # Take marks for five subjects

    print("---5 subject marks---")
    for i in range(5):
        while True:
            try:
                # Get marks from the user
                mark = float(input(f"Enter marks for subject {i + 1}: "))
                # Add the mark to the list
                marks.append(mark)
                break
            except ValueError:
                print("Invalid input! Please enter a number")

    # Calculate average marks
    avg = sum(marks) / len(marks)
    ## Display average, highest, and lowest marks
    print(f"\nAverage marks: {avg}")
    print("Highest marks: ", max(marks))
    print("Lowest marks: ", min(marks))

    # Display the sorted list
    marks.sort(reverse=True)
    print("\nMarks in decending order: ", marks)

# Calling the function
manage_marks()
```
