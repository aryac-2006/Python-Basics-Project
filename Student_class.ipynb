```python
class Student:
    # Constructor to initialize student details
    def __init__(self, name, roll_no):
        self.stu_name = name
        self.stu_rollno = roll_no
        self.marks_list = []

    # Method to add marks
    def add_marks(self, marks):
        if 0 <= marks < 100:
            self.marks_list.append(marks)
        else:
            raise ValueError("Invalid marks")

    # Method to calculate average marks
    def get_average(self):
        return sum(self.marks_list) / len(self.marks_list)

    # Method to display details of student
    def display_info(self):
        print("\n===Student Details===")
        print("Name:", self.stu_name)
        print("Roll No:", self.stu_rollno)
        print("Marks:", self.marks_list)
        print("Average Marks:", self.get_average())

name = input("Enter Student Name: ")
roll_no = input("Enter Student Roll No: ")

stu = Student(name, roll_no)
try:
    n = int(input("\nHow many marks do you want to enter? : "))
    for i in range(n):
        try:
            marks = float(input("Enter marks: "))
            stu.add_marks(marks)

        except ValueError as e:
             print(e)
except ValueError:
    print("Invalid input! please enter a valid integer")
stu.display_info()
```
