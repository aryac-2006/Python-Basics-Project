```python
import math
import random

history = {}

def basic_arithmetic():
    try:
        num1 = int(input("Enter first number: "))
        num2 = int(input("Enter second number: "))
        print("\n1). Addition")
        print("2). Subtraction")
        print("3). Multiplication")
        print("4). Division")

        choice = int(input("Choice Option: "))

        if choice == 1:
            result = num1 + num2
        elif choice == 2:
            result = num1 - num2
        elif choice == 3:
            result = num1 * num2
        elif choice == 4:
            result = num1 / num2
        else:
            print("Invalid Choice")

        print("Result: ", result)
        return result

    except ZeroDivisionError:
        print("Cannot Divide by Zero")
    except ValueError:
        print("Invalid Input")

def scientific_calculation():
    try:
        num = float(input("Enter a number: "))
        print("\n1). Square root")
        print("2). Power")
        print("3). Factorial")

        choice = int(input("Choice Option: "))
        if choice == 1:
            result = math.sqrt(num)
        elif choice == 2:
            power = float(input("Enter a power: "))
            result = math.pow(num, power)
        elif choice == 3:
            result = math.factorial(int(num))
        else:
            print("Invalid Choice")

        print("Result: ", result)
        return result

    except ValueError:
        print("Invalid Input")

def random_numbers():
    numbers = []
    for i in range(5):
        numbers.append(random.randint(1, 100))
    print("Random Number: ", numbers)
    return numbers

while True:
        print("\n----Smart Calculator & Data Manager----")
        print("1. Basic Arithmetic")
        print("2. Scientific Calculation")
        print("3. Generate Random Numbers")
        print("4. Store result")
        print("5. View History")
        print("6. Exit")

        choice = int(input("Choice Option: "))
        if choice == 1:
            result = basic_arithmetic()
        elif choice == 2:
            result = scientific_calculation()
        elif choice == 3:
            result = random_numbers()
        elif choice == 4:
            if 'result' in locals():
                timestamp = input("Enter a timestamp (Example: 29-06-2026 10:30 AM):")
                history[timestamp] = result
                print("Result stored successfully")
            else:
                print("No Result available")
        elif choice == 5:
            if history:
                print("\nHistory:")
                for time, value in history.items():
                    print(time, ": ", value)
            else:
                print("No History:")
        elif choice == 6:
            print("Thank you!")
            break
        else:
            print("Invalid Choice")

```
